```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
path=r"/content/archive (13).zip"
data=pd.read_csv(path,compression="zip")
print(data.shape)
print(data.head())
```

```python
print(data.dtypes)
print(data.isnull().sum())
```

```python
object_data=data.select_dtypes(include=["object"]).columns
data[object_data]=data[object_data].fillna('missing')

print(data.isnull().sum())
```
